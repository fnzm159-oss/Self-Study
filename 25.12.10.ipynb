```python
# 별 찍기_1

star = int(input("숫자를 입력: "))  # 숫자 입력 받기
a = 1  # a를 설정해서 while의 반복문 탈출 만들기
b = 0  # b를 이용해서 *의 개수 짝기

while a <= star:
    a += 1
    b += 1
    print('*' * b)
```

```python
# 별 찍기_2

star = int(input("숫자를 입력: "))  # 숫자 입력 받기
a = 1  # a를 설정해서 while의 반복문 탈출 만들기
b = star  # star의 갯수 만큼 입력을 받고 빈칸을 하나씩 빼기
c = 0   # c를 이용해서 *의 개수 찍기

while a <= star:
    a += 1
    b -= 1
    c += 1
    print(' ' * b, '*' * c)
```
