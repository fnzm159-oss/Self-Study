```python
import numpy as np
```

```python
np.arange(2, 1000, 2)   # 앞에 두자리 범위 지정, 세번째 자리는 구간을 설정
```

```python
np.arange(0, 1, 0.1)  # 앞에 두자리 범위 지정, 세번째 자리는 구간을 설정
```

```python
data = [0, 1, 2, 3, 4, 5]

df = np.array(data)   # data를 넘파이의 배열로 설정

df.reshape(2,3)       # 행과 열을 설정
```

```python
df.flatten()    # 1차원으로 펼치는 flatten() 함수를 사용
```
