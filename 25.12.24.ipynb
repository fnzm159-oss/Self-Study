```python
# 체스 킹, 퀸, 룩, 비숍, 나이트, 폰 개수 맞추기


perfect = [1, 1, 2, 2, 2, 8]                   # 완벽한 체스 세트

having = list(map(int, input().split()))       # 내가 갖고 있는 체스를 입력

for i in range (6):                            # 리스트별로 같은 순서로 꺼내서 비교하므로 리스트 안에 갯수를 반영하여 적용(리스트 순서 0-5니까 6을 적음)
    print(perfect[i] - having[i], end=' ')     # perfect에서 having을 빼고 사이에 공백을 넣어서 출력
```

```python
# 팰린드롬인지 확인하기

text = input()

if text[:] == text[::-1]:     # 원래 리스트의 순서와 리버스한 리스트의 순서를 비교
    print(1)                  # 값이 같으면 1을 출력 (팰린드롬 O)
else:
    print(0)                  # 값이 다르면 0을 출력 (팰린드롬 X)
```
