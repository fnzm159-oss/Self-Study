```python
#[조건문 + 변수를 리턴하는 함수]

n = int(input())

def result(n):
    if (n % 3 == 0) and (n % 5 == 0):
        return "FizzBuzz"
    elif n % 3 == 0:
        return "Fizz"
    elif n % 5 == 0:
        return "Buzz"
    else:
        'None'

print(result(n))
```

```python
#[조건문 + 값을 리턴하는 함수]

a = 2
b = 2
c = 3

def result(a,b,c):
    if a == b == c:
        return a * b * c
    elif a != b != c:
        return max(a, b, c)
    else:
        if a == b:
            return c
        elif a == c:
            return b
        else:
            return a
        
print(result(a,b,c))
```
