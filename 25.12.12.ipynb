```python
# 정수 N개로 이루어진 수열 A와 정수 X가 주어진다. 이때 A에서 X 보다 작은 수를 모두 출력하는 프로그램을 작성하시오.

N, X = map(int, input().split())
A = list(map(int, input().split()))
result = []

# A 속애 있는 값을 X와 비교하여 작다면 result에 추가
for i in A:
    if i < X:
        result.append(i)
print(*result) # *을 입력해서 리스트 안에 값만 출력

```

```python
# N개의 정수가 주어진다. 이때 최솟값과 최댓값을 구하는 프로그램을 작성하시오.

N = int(input())
A = list(map(int, input().split()))

print(min(A), max(A))
```
