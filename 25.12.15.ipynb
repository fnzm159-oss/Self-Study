```python
# [과제 제출 안 한 사람 찾기]

number = [*range(1,31)]  # 총 인원 30명을 생성

student = [] # 제출한 인원을 추가할 리스트를 생성

no = []  # 제출하지 않은 인원을 추가할 리스트를 생성

for i in range(1,29):
    num = int(input())  # 28명의 과제 제출 인원을 입력
    student.append(num) # stduent 리스트에 넣기

for j in number:
    if j not in student:  # 총 30번 까지의 번호 중 제출한 인원(studnet)에 포함되지 않은 번호를 찾아냄
        no.append(j)

print(*no, sep='\n')  # 리스트 형식 출력를 *로 바꾸고, sep으로 줄바꿈을 실행


```

```python
# 과제: 나머지 구하기
# 나의 구상: 
# 1) 반복문 input으로 리스트로 입력을 받고
# 2) 다른 갯수는 append로 카운트를 올리고

number = []

a = []

b = []

for i in range(1,11):   # 숫자 10개를 입력
    num = int(input())
    number.append(num)

for j in number:  # 숫자를 42로 나눈 나머지를 a리스트에 추가
    a.append(j%42)

for k in a:    # b에 k가 들어있지 않으면 append를 실시 => 중복은 패스를 함
    if k not in b:
        b.append(k)

print(len(b))
```
