```python
N = int(input("정수를 입력: "))

list = []

A = []
B = []
C = []
D = []

a = 1

while a <= N:
    i = int(input("점수를 입력: "))
    a += 1
    if 0 <= i <= 100:
        list.append(i)
print(list)


for score in list:
    if score >= 90:
        A.append(list)
    elif 70 <= score < 90:
        B.append(list)
    elif 40 <= score < 70:
        C.append(list)
    else:
        D.append(list)

print(f"A:{len(A)}, B:{len(B)}, C:{len(C)}, D:{len(D)}")


mmax = max(len(A), len(B), len(C), len(D))


if (len(A) == len(B)) or (len(A) == len(C)) or (len(A) == len(D)):   # Tie를 위에 얹어놔야 필터링 가능
    print('Tie')
elif (len(B) == len(C)) or (len(B) == len(D)):
    print('Tie')
elif (len(A) == len(B)):
    print('Tie')
elif mmax == len(A):
    print('A')
elif mmax == len(B):
    print('B')
elif mmax == len(C):
    print('C')
elif mmax == len(D):
    print('D')
```
