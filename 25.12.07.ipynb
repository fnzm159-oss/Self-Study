```python
#[1에서부터 6까지의 눈을 가진 3개의 주사위를 던져서 규칙에 따라 상금을 받는 게임]

A, B, C = map(int, input().split())

if A==B==C:
    print(10000+A*1000)
elif (A==B) or (A==C):
    print(1000+A*100)
elif (B==C):
    print(1000+B*100)
else:
    print(max(A,B,C)*100)
```
