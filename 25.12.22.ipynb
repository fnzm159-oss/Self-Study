```python
# 문자열 S를 입력 받은 후에, 각 문자를 R번 반복해 새 문자열 P를 만든 후 출력하는 프로그램을 작성하시오.
# 첫 번째 문자를 R번 반복하고, 두 번째 문자를 R번 반복하는 식으로 P를 만들면 된다. S에는 QR Code "alphanumeric" 문자만 들어있다.

S = input()
R = int(input())

a = []


for i in S:        # 입력한 문잗열 S를 i에 넣기
    a.append(i*R)  # S의 문자열을 하나씩 꺼내서 R만큼 곱해서 리스트 a에 추가

print(*a, sep="")  # *을 통해 언패킹을 진행 후 공백을 없애기 위해 sep을 사용
```

```python
# 영어 대소문자와 공백으로 이루어진 문자열이 주어진다. 이 문자열에는 몇 개의 단어가 있을까? 이를 구하는 프로그램을 작성하시오.
# 단, 한 단어가 여러번 등장하면 횟수만큼 모두 세어야 한다.

sen = input()  

a = 1             # 단어의 갯수는 공백 수 +1 이므로 a의 초기 설정을 1로 함

for i in sen:     # i에 sen을 넣기
    if i == " ":  # 공백을 필터링
        a += 1    # 공백을 만났다면 a의 숫자를 하나씩 올림

print(a)
```
