```python
# [오븐 시계 _ 소요시간 입력 하면 끝나는 시간을 출력]

A = int(input("시: "))
B = int(input("분: "))
C = int(input("분: "))

if (0 <= A <= 23) and (0<= B <= 59) and (0 <= C <= 1000):
    if ((B + C) % 60 >= 0) and ((A+(B+C)//60) < 24):
        print(f"{A+(B+C)//60} {(B+C)%60}")
    elif ((B + C) % 60 >= 0) and ((A+(B+C)//60) >= 24):
         print(f"{(A+(B+C)//60)-24} {(B+C)%60}")

    else:
        print(f"{A} {B+C}")
```
