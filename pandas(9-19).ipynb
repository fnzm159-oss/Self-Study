```python
import pandas as pd
```

```python
#[9] 데이터를 로드하라. 컬럼이 한글이기에 적절한 처리해야함
Dataurl = 'https://raw.githubusercontent.com/Datamanim/pandas/main/Jeju.csv'
df = pd.read_csv(Dataurl, encoding='euc-kr')
type(df)
```

```python
#[10] 데이터 마지막 3개행을 출력
df.tail(3)
```

```python
#[11] 수치형 변수를 가진 컬럼을 출력  // number <-> object  // exclude <-> include
df.select_dtypes(exclude=object).columns
```

```python
#[12] 범주형 변수를 가진 컬럼을 출력
df.select_dtypes(include=object).columns
```

```python
#[13] 각 컬럼의 결측치 숫자를 파악하라
df.isna().sum()
```

```python
#[14] 각 컬럼의 데이터 수, 데이터 타입을 한번에 확인하라
df.info()
```

```python
#[15] 각 수치형 변수의 분포(사분위, 평균, 표준편차, 최대, 최소)를 확인하라.
df.describe()
```

```python
#[16] 거주인구 컬럼의 값들을 출력하라
df['거주인구']
```

```python
#[17] 평균 속도 컬럼의 4분위 범위(IQR) 값을 구하라
df['평균 속도'].quantile(0.75) - df['평균 속도'].quantile(0.25)
```

```python
#[18] 읍면동명 컬럼의 유일값 갯수를 출력하라
df['읍면동명'].nunique()
```

```python
#[19] 읍면동명 컬럼의 유일값을 모두 출력하라
df['읍면동명'].unique()
```
