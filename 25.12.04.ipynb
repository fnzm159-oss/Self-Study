```python
# [사분면 고르기]
x = int(input("x값을 입력: "))
y = int(input("y값을 입력: "))

if (-1000 <= x <= 1000 and x != 0) and (-1000 <= y <= 1000 and y != 0):
    if (x>0) and (y>0):
        print(1)
    elif (x<0) and (y>0):
        print(2)
    elif (x<0) and (y<0):
        print(3)
    elif (x>0) and (y<0):
        print(4)
```

```python
# [알람 시계 _ 45분 일찍 알람 설정하기]
H = int(input("시: "))
M = int(input("분: "))

if (0 <= H <= 23) and (0 <= M <= 59):
    if M - 45 < 0:
        if H == 0:
            print(f"{H+23} {M+15}")
        elif H != 0:
            print(f"{H-1} {M+15}")
    else:
        print(f"{H} {M-45}")
```
