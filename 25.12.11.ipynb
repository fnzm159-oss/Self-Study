```python
# 두 정수 A와 B를 입력받은 다음, A+B를 출력하는 프로그램을 작성하시오

while True:
    A, B = map(int,input().split())
    if A == 0 and B ==0:
        break
    print(A+B)
```

```python
# 총 N개의 정수가 주어졌을 때, 정수 v가 몇 개인지 구하는 프로그램을 작성하시오 /// 자꾸 막혀서 제미나이와 토의

N = int(input())
a = input().split() # 일단 입력은 다 받아서 임시 저장
num = []               # 결과를 담을 빈 리스트

for i in range(N):     # 0부터 N-1까지, 딱 N번만 반복
    num.append(int(a[i]))

v = int(input())
print(num.count(v))
```
