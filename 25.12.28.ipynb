```python
import numpy as np
arr = np.arange(3)
arr.reshape(3,1)    # 세개의 행과 하나의 열을 보기 위한 reshape
```

--> 현재 존재하는 값은 6개인데, 2X4의 배열로 만들려면 8개의 값이 필요하기 때문에 에러가 남

```python
np.arange(6).reshape(2,3)
```

```python
arr = np.ones(20, dtype=int).reshape(4,5)  # 1이 20개 들어있는 1차원 배열 생성 --> 4X5로 reshape
print(arr)
```
