```python
# 알파벳 대소문자로 된 단어가 주어지면, 이 단어에서 가장 많이 사용된 알파벳이 무엇인지 알아내는 프로그램을 작성하시오. 단, 대문자와 소문자를 구분하지 않는다.


# 단어끼리의 비교를 하기 위해서 모두 대문자로 변환
sen = input().upper()


# 중복 문자를 제거하여 문자별로 분류
set_sen = list(set(sen))


# 원래 문자와 중복 문자를 비교한 것을 count를 통해 갯수를 파악
num_set_sen = []
for i in set_sen:
    num_set_sen.append(sen.count(i))

# 최대 count 찾기
max_count = max(num_set_sen)

# 최대 count가 중복이 있는지 확인
if num_set_sen.count(max_count) > 1:          # 최대 count를 가진 단어가 여러개면 ?를 출력
    print('?')
else:
    max_word = num_set_sen.index(max_count)   # 최대 count의 인덱스를 찾기
    print(set_sen[max_word])                  # 해당 인덱스의 단어를 출력
```
