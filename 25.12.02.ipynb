```python
n = int(input("리스트의 크기를 지정하시오: "))
numbers = []
even = 0
odd = 0

for i in range(n):
    num = int(input(f"{i+1}번째 입력"))
    numbers.append(num)

for j in numbers:
    if j % 2 == 0:
        even += 1
    else:
        odd += 1

print(f"짝수 개수: {even}")
print(f"홀수 개수: {odd}")
```
