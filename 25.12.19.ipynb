```python
# [알파벳 소문자로만 이루어진 단어 S가 주어진다. 각각의 알파벳에 대해서, 단어에 포함되어 있는 경우에는 처음 등장하는 위치를,\
# 포함되어 있지 않은 경우에는 -1을 출력하는 프로그램을 작성하시오.]


# 내가 처음 짠 코드
# 1. S[a]의 문제: a는 0부터 25까지 증가하지만, 입력받은 단어 S가 만약 5글자라면 S[5]부터는 IndexError가 발생
# 2. 비교 대상의 오류: S[a] == i는 "단어의 a번째 글자가 i인가?"를 묻는 것이지, "단어 전체에 i가 포함되어 있는가?"를 묻는 것이 아님
# 3. list 예약어: 파이썬에서 list는 리스트를 만드는 내장 함수 이름이므로, 변수명은 result_list 등으로 바꾸는 것이 안전


import string

S = input()

a = 0

list = []

lower = string.ascii_lowercase


for i in lower:
    if S[a] == i:
        list.append(S.find(i))
    elif S[a] != i:
        list.append(-1)
    elif a == 25:
        break
    a += 1

print(list)
```

```python
# 내가 짜려고 했던 if/else 유지하는 방법(from.Gemini)

import string
S = input()
result_list = []
lower = string.ascii_lowercase

for i in lower:
    if i in S:     # 내가 하려고 했던 S안에 i가 있는지 확인하는게 아니라 반대로 i안에 S가 있는지를 확인하는 방법을 사용
        result_list.append(S.find(i))
    else:
        result_list.append(-1)

print(*(result_list))
```

```python
# 더 효율적인 방법 find의 특성 활용 (from.Gemini)

import string

S = input()
lower = string.ascii_lowercase

for i in lower:
    print(S.find(i), end=' ')    # find는 해당 문자 없으면 자동으로 -1을 반환하는 기능이 내재
```
