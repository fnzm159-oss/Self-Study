```python
# 두 수가 주어졌을 때 숫자를 거꾸로 읽는 상수의 대답을 출력하는 프로그램을 작성

A, B = map(int, input().split())

if str(A)[::-1] > str(B)[::-1]:     # 입력받은 수를 str()로 문자열로 바꾸고 [::-1]을 통해 반전을 시켜줌 --> 크기 비교를 통해 더 큰 것을 출력
    print(str(A)[::-1])
else:
    print(str(B)[::-1])
```

```python
# 딕셔너리를 이용하는 것은 같음
# 각 알파벳에 전화를 거는데 필요한 시간을 value로 넣음 --> value는 각 알파벳이 해당하는 숫자 +1을 넣어줌 (예, 다이얼에서 A는 2에 위치하므로 value는 3을 넣음)

time = {
    'A': 3, 'B': 3, 'C': 3,
    'D': 4, 'E': 4, 'F': 4,
    'G': 5, 'H': 5, 'I': 5,
    'J': 6, 'K': 6, 'L': 6,
    'M': 7, 'N': 7, 'O': 7,
    'P': 8, 'Q': 8, 'R': 8, 'S': 8,
    'T': 9, 'U': 9, 'V': 9,
    'W': 10, 'X': 10, 'Y': 10, 'Z': 10
}


word = input()

total_time = 0              # 총 시간을 0으로 설정

for i in word:              # i에 입력한 단어를 넣음
    total_time += time[i]   # 단어 하나씩 들어가면서 각 알파벳에 해당하는 숫자가 toal에 더해짐

print(total_time)
```
