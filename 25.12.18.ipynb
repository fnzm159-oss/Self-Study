```python
#[문자열을 입력으로 주면 문자열의 첫 글자와 마지막 글자를 출력하는 프로그램을 작성하시오.]

n = int(input())     # 테스트 케이스의 개수를 입력
a = 0


while a <= n:   # 테스트 케이스 개수 만큼 입력을 받을 수 있게 반복문을 사용
    a += 1
    if a == 1:
        i = (input())
    elif a == 2:
        j = (input())
    elif a == 3:
        k = (input())

print(i[0] + i[-1])    # 첫 글자와 마지막 글자를 출력
print(j[0] + j[-1])
print(k[0] + k[-1])
```

```python
#[N개의 숫자가 공백 없이 쓰여있다. 이 숫자를 모두 합해서 출력하는 프로그램을 작성하시오.]

N = int(input())   # N개를 입력

num = map(int, input())  # 공백 없이 숫자를 입력

num_list = [int(i) for i in num]  # 리스트컴프리헨션을 이용해 공백 없는 숫자를 한자리씩 분류 

print(sum(num_list))
```
