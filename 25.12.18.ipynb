```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/Datamanim/pandas/main/AB_NYC_2019.csv")
```

```python
#[53] neighbourhood 값과 neighbourhood_group 값에 따른 price의 평균을 계층적 indexing 없이 구하고 nan 값은 -999값으로 채워라
Ans = df.groupby(['neighbourhood','neighbourhood_group'])\
    .price.mean()\
    .unstack()\
    .fillna(-999)
Ans.head()
# 그룹핑을 진행
# 값의 평균을 구함
# 열을 늘어진 데이터를 행으로 펼치기
# 빈칸을 원하는 숫자로 채우기
```

```python
#[54] 데이터 중 neighbourhood_group 값이 Queens값을 가지는 데이터들 중 neighbourhood 그룹별로 price값의 평균, 분산, 최대, 최소값을 구하라
Ans = df[df.neighbourhood_group == 'Queens']\
        .groupby(['neighbourhood'])\
        .price.agg(['mean', 'var', 'max', 'min'])
Ans.head()
# Queens 값을 갖는 neighbourhood_group을 분류
# neighbourhood로 그룹핑
# agg를 이용해서 price를 분류
```

```python
#[55] 데이터 중 neighbourhood_group값에 따른 room_type 컬럼의 숫자를 구하고 neighbourhood_group값을 기준으로 각 값의 비율을 구하여라
df.groupby('neighbourhood_group')['room_type']\
    .value_counts(normalize=True).unstack()

# 그룹핑을 하고 ['room_type']의 정보만 꺼내옴
# normalize로 비율을 가져옴
```
