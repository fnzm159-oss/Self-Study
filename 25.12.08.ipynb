```python
# [n이 주어졌을 때 1부터 n까지 합을 구하는 프로그램]

n = int(input("숫자를 입력: "))
t = 0

for i in range(1,n+1):
    t += i
print(t)
```

```python
# [두 정수 A와 B를 입력받은 다음, A+B를 출력하는 프로그램을 작성하시오.]

T = int(input("케이스 개수를 입력: "))

for i in range(1,T+1):
    A, B = map(int, input().split())
    print(f"Case #{i}: {A} + {B} = {A+B}")
```
