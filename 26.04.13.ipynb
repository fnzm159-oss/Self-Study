#### 1번 _ 풀이

```python
# 인덱스 접근이 아니라 map을 이용하기 때문에 값으 처음에 다 부여하기
# 줄바꿈 함수 -->  sep='\n'

A, B, C = map(int, input().split())

print((A+B)%C,
      ((A%C)+(B%C))%C,
      (A*B)%C,
      ((A%C)*(B%C)%C),
      sep='\n')
```
